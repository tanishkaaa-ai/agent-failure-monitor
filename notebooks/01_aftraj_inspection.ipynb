# AFTraj-2K Dataset Inspection

This notebook inspects the **AFTraj-2K** dataset downloaded from Hugging Face ().
It explores schema, trajectory composition, turn-level structures, and failure annotations.

```python
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load safe and unsafe splits
safe_df = pd.read_parquet("../data/raw/aftraj_safe.parquet")
unsafe_df = pd.read_parquet("../data/raw/aftraj_unsafe.parquet")
with open("../data/raw/splits_test.json") as f:
    splits = json.load(f)

print(f"Safe trajectories:   {len(safe_df)}")
print(f"Unsafe trajectories: {len(unsafe_df)}")
print(f"Total trajectories:  {len(safe_df) + len(unsafe_df)}")
```

## 1. Schema & Column Overview

```python
print("Safe Columns:", list(safe_df.columns))
print("Unsafe Columns:", list(unsafe_df.columns))
diff_cols = set(unsafe_df.columns) - set(safe_df.columns)
print("
Additional failure annotation columns in unsafe:", list(diff_cols))
```

## 2. Domain Distribution & Unsafe Failure Sources

```python
domain_df = pd.DataFrame({
    "Safe": safe_df["domain"].value_counts(),
    "Unsafe": unsafe_df["domain"].value_counts()
}).fillna(0).astype(int)
domain_df["Total"] = domain_df["Safe"] + domain_df["Unsafe"]
print(domain_df)
print("
Unsafe failure sources (injected vs naturally diagnosed):
", unsafe_df["unsafe_source"].value_counts())
```

## 3. Decisive Error Step (^*$) Distribution

```python
print(unsafe_df["mistake_step"].describe())
print("
Top Responsible Agents for Mistakes:
", unsafe_df["mistake_agent"].value_counts().head(10))
```
